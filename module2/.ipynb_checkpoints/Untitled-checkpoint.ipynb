## Первая эпоха
Считаем расстояния:
#### Нейрон 1:
X1: $\sqrt{(3-(-0.04))^2+(0-0.08)^2} = 3.041052449399714$
\
X2 $\sqrt{(1-(-0.04))^2+(5-0.08)^2} = 5.028717530345088$
\
X3 $\sqrt{(3-(-0.04))^2+(4-0.08)^2} = 4.960645119336799$
\
X4 $\sqrt{(8-(-0.04))^2+(1-0.08)^2} = 8.092465631684819$

#### Нейрон 2:
X1: $\sqrt{(3-(1.21))^2+(0-1.15)^2} = 2.127580785775243$
\
X2 $\sqrt{(1-(1.21))^2+(5-1.15)^2} = 3.8557230190977156$
\
X3 $\sqrt{(3-(1.21))^2+(4-1.15)^2} = 3.3655014485214534$
\
X4 $\sqrt{(8-(1.21))^2+(1-1.15)^2} = 6.791656646209377$

```python
def r(a,b):
    return ((a[0]-b[0])**2 + (a[1]-b[1])**2)**0.5
```

```python
import math
df = [(3,0), (1,5), (3,4), (8,1)]
weights = [[-0.04, 0.08], [1.21, 1.15]]
for i,w in enumerate(weights):
    for j, x in enumerate(df):
        print(f'{i+1}-й нейрон, Х{j+1}: {r(x,w)}')
    print()
```

### Нейрон 2 победитель, при чем для всех
обновляем его веса

```python
new_weights = weights[1]
lr = 0.5
for j, x in enumerate(df):
    print(f'X{j+1}')
    print(f'w1 = {new_weights[0]} - 0.5 * ({new_weights[0]} - {x[0]})')
    print(f'w2 = {new_weights[1]} - 0.5 * ({new_weights[1]} - {x[1]})')
    print()
    new_weights[0] -= lr*(new_weights[0] - x[0])
    new_weights[1] -= lr*(new_weights[1] - x[1])
print((new_weights))
```

#### $[5.3836328125, 2.2623046875]$ - Новые веса нейрона 2

## Вторая эпоха

```python
import math
weights[1] = new_weights
for i,w in enumerate(weights):
    for j, x in enumerate(df):
        print(f'{i+1}-й нейрон, Х{j+1}: {r(x,w)}')
    print()
```

#### В 1, 2 точках победитель 1 нейрон, в 3,4 - 2 нейрон

```python
new_weights = weights.copy()

lr = 0.5

for j, x in enumerate(df):
    
    neuron = 0 if j < 2 else 1
    print(f'{neuron+1}-й Нейрон меняем:')
    
    print(f'X{j+1}')
    
    print(f'w1 = {new_weights[neuron][0]} - 0.5 * ({new_weights[neuron][0]} - {x[0]})')
    print(f'w2 = {new_weights[neuron][1]} - 0.5 * ({new_weights[neuron][1]} - {x[1]})')
    
    print()
    
    new_weights[neuron][0] -= lr*(new_weights[neuron][0] - x[0])
    new_weights[neuron][1] -= lr*(new_weights[neuron][1] - x[1])

print((new_weights))
```

#### $[1.24, 2.52], [6.095908203125, 2.065576171875]]$ - новые веса нейронов

```python
for x in df:
    print(r(new_weights[0],x), r(new_weights[1],x))
```
