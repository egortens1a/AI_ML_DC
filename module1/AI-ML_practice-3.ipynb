Для заданного нечеткого множества А найти: носитель, ядро, высоту, множества уровня для заданных значений α.

Носитель нечеткого множества - множество элементов, пренадлежащих нечеткому множеству хотя бы в какой то ненулевой степени

Ядро нечеткого множества - множество элементов, пренадлежность которых равна 1

Высота нечеткного множества - максимальная степень пренаддлежности

Множества уровня для заданных значений alpha  - срез множества - множество, где пренадлежность элементов >=alpha (или >alpha, если строгое)
Нормальное множество - высота = 1

Субнормальное множество - высота < 1 (можно нормализовать, домножив все степени пренадлежности на столько, сколько надо максимальной степени пренадлежности чтобы стать 1

Унимодальное множество - мощность ядра = 1
Объединение двух НМ - НЧ, где взяты максимальные степени пренадлежности каждого элемента из множеств
Пересечение двух НМ - НМ, гдк взяты минимальные степени пренадлежности каждого элемента из множеств
Отрицание НМ - степень пренадлежности эл = 1 - (степень пренадлежности элемента)
